## Check that embeddings are correctly saved


```python
from pathlib import Path
import torch

embedding_path = Path(
    "../data/ucf101_embeddings/test/ApplyEyeMakeup/v_ApplyEyeMakeup_g01_c02.pt"
)

if not embedding_path.exists():
    raise FileNotFoundError(f"Embedding file not found: {embedding_path}")

video_embeddings = torch.load(embedding_path)

print(f"Shape of video embeddings: {video_embeddings.shape}")
print(f"Data type: {video_embeddings.dtype}")

print("First 5 frame embeddings:")
print(video_embeddings[:5])
```
